```python
conditions = [
    (df["Rainfall (mm)"] < 7.6),
    (df["Rainfall (mm)"] >= 7.6) & (df["Rainfall (mm)"] < 35.5),
    (df["Rainfall (mm)"] >= 35.5) & (df["Rainfall (mm)"] < 64.5),
    (df["Rainfall (mm)"] >= 64.5) & (df["Rainfall (mm)"] < 124.4),
    (df["Rainfall (mm)"] >= 124.4)
]
choices = ["minimum risk", "low", "mild", "warning", "high"]
df["Risk"] = np.select(conditions, choices, default="unknown")
```

```python
df["Risk"].value_counts()
```

```python
df['Risk'] = pd.Categorical(df['Risk'], categories=choices,ordered=True)
df['Risk_encoded'] = df['Risk'].cat.codes
```

```python
df.drop(columns=['Risk'],inplace=True)
```
